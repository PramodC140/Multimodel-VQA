```python


# Helper function to load DAQUAR images (assuming they're stored in NYU Depth V2 format)
def load_nyu_depth_images(image_dir):
    """Load and cache NYU-Depth images"""
    logger.info(f"Loading images from {image_dir}")
    image_data = {}
    
    # Load all images from the directory
    for img_file in os.listdir(image_dir):
        if img_file.endswith(('.jpg', '.png', '.jpeg')):
            img_id = os.path.splitext(img_file)[0]
            img_path = os.path.join(image_dir, img_file)
            try:
                image = Image.open(img_path).convert('RGB')
                image_data[img_id] = image
            except Exception as e:
                logger.warning(f"Failed to load image {img_file}: {str(e)}")
                continue
    
    return image_data

# Load answer space
def load_answer_space(answer_space_path):
    """Load the answer space from file"""
    with open(answer_space_path, 'r') as f:
        answers = [line.strip() for line in f.readlines()]
    
    # Create mappings
    answer_to_idx = {answer: idx for idx, answer in enumerate(answers)}
    idx_to_answer = {idx: answer for answer, idx in answer_to_idx.items()}
    
    return answers, answer_to_idx, idx_to_answer
```

```python
class DAQUARDataset(Dataset):
    def __init__(self, csv_file, answer_to_idx, image_data, tokenizer, transform=None, max_length=128, is_train=True):
        """
        Initialize DAQUARDataset
        
        Args:
            csv_file (str): Path to the CSV file relative to config.DATA_PATH
            answer_to_idx (dict): Mapping from answer string to index
            image_data (dict): Dictionary of image data
            tokenizer: BERT tokenizer instance
            transform: Image transform pipeline
            max_length (int): Maximum sequence length for tokenizer
            is_train (bool): Whether this is training set
        """
        logger.info(f"Initializing {'training' if is_train else 'validation'} dataset from {csv_file}")
        
        self.df = pd.read_csv(os.path.join(config.DATA_PATH, csv_file))
        self.answer_to_idx = answer_to_idx
        self.image_data = image_data
        self.tokenizer = tokenizer
        self.transform = transform or transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor()
        ])
        self.max_length = max_length
        self.is_train = is_train
        
        # Initialize question types and indices for balanced sampling
        if self.is_train:
            self.question_types = self.extract_question_types()
            self.question_type_indices = self.group_by_question_type()
            logger.info(f"Found {len(self.question_type_indices)} question types")
            
        logger.info(f"Dataset initialized with {len(self.df)} samples")
    
    def extract_question_types(self):
        """Extract question types based on first word"""
        question_types = []
        for q in self.df['question']:
            try:
                first_word = q.lower().split()[0]
                question_types.append(first_word)
            except (AttributeError, IndexError) as e:
                logger.warning(f"Error processing question: {q}. Error: {str(e)}")
                question_types.append("unknown")
        return question_types
    
    def group_by_question_type(self):
        """Group indices by question type for balanced sampling"""
        groups = {}
        for i, qtype in enumerate(self.question_types):
            if qtype not in groups:
                groups[qtype] = []
            groups[qtype].append(i)
        return groups
    
    def get_balanced_sample_idx(self):
        """Get a sample with balanced question types"""
        if not self.is_train:
            return random.randint(0, len(self.df) - 1)
        # Randomly select a question type, then a sample from that type
        qtype = random.choice(list(self.question_type_indices.keys()))
        return random.choice(self.question_type_indices[qtype])
    
    def __len__(self):
        return len(self.df)
    
    def preprocess_image(self, image_id):
        """Process image with error handling"""
        try:
            image = self.image_data[image_id]
            if isinstance(image, np.ndarray):
                image = Image.fromarray(image.astype(np.uint8))
            return self.transform(image)
        except Exception as e:
            logger.warning(f"Error processing image {image_id}: {str(e)}")
            default_image = Image.fromarray(np.zeros((224, 224, 3), dtype=np.uint8))
            return self.transform(default_image)
    
    def preprocess_question(self, question):
        """Process question text with prefixes"""
        if not self.is_train or random.random() >= 0.5:
            return question
            
        question_prefixes = {
            "what": "[WHAT]",
            "how": "[HOW]",
            "where": "[WHERE]",
            "which": "[WHICH]",
            "who": "[WHO]",
            "is": "[YES_NO]",
            "are": "[YES_NO]"
        }
        
        try:
            first_word = question.lower().split()[0]
            prefix = question_prefixes.get(first_word, "")
            return f"{prefix} {question}" if prefix else question
        except (AttributeError, IndexError) as e:
            logger.warning(f"Error processing question: {question}. Error: {str(e)}")
            return question
    
    def process_answer(self, answer):
        """Process answer with error handling"""
        if not answer or pd.isna(answer):
            return 0
            
        try:
            answer = str(answer).strip().lower()
            if ',' in answer:
                answer = answer.split(',')[0].strip()
            return self.answer_to_idx.get(answer, 0)
        except Exception as e:
            logger.warning(f"Error processing answer '{answer}': {str(e)}")
            return 0
    
    def __getitem__(self, idx):
        if self.is_train and random.random() < 0.2:  # 20% chance of balanced sampling
            idx = self.get_balanced_sample_idx()
        
        try:
            row = self.df.iloc[idx]
            
            # Process image
            image_id = row['image_id']
            image = self.preprocess_image(image_id)
            
            # Process question
            question = self.preprocess_question(row['question'])
            
            # Tokenize question
            encoded_text = self.tokenizer(
                question,
                padding='max_length',
                max_length=self.max_length,
                truncation=True,
                return_tensors='pt'
            )
            
            # Process answer
            label = self.process_answer(row.get('answer', ''))
            
            return {
                'image': image,  # Changed from 'pixel_values' to 'image'
                'input_ids': encoded_text['input_ids'].squeeze(0),
                'attention_mask': encoded_text['attention_mask'].squeeze(0),
                'answer': torch.tensor(label, dtype=torch.long),  # Changed from 'label' to 'answer'
                'question_text': question,  # Added for debugging
                'image_id': image_id,  # Added for debugging
            }
            
        except Exception as e:
            logger.error(f"Error processing item {idx}: {str(e)}")
            logger.error(traceback.format_exc())
            # Return a default item in case of error
            default_image = torch.zeros((3, 224, 224))
            default_text = self.tokenizer(
                "error",
                padding='max_length',
                max_length=self.max_length,
                truncation=True,
                return_tensors='pt'
            )
            return {
                'image': default_image,
                'input_ids': default_text['input_ids'].squeeze(0),
                'attention_mask': default_text['attention_mask'].squeeze(0),
                'answer': torch.tensor(0, dtype=torch.long),
                'question_text': "error",
                'image_id': -1,
            }
```

```python
from transformers import ViTConfig, ViTModel
from transformers import ViTConfig, ViTModel

class AdvancedVisionFeatures(nn.Module):
    def __init__(self, config):
        super().__init__()
        # Load and modify ViT config
        vit_config = ViTConfig.from_pretrained(config.VIT_MODEL)
        vit_config.gradient_checkpointing = config.USE_GRADIENT_CHECKPOINTING
        
        # Initialize ViT model with mixed precision support
        self.vit = ViTModel.from_pretrained(
            config.VIT_MODEL,
            config=vit_config,
            add_pooling_layer=False,
            torch_dtype=torch.float16 if config.USE_AMP else torch.float32
        )
        
        hidden_size = vit_config.hidden_size

        # Multi-scale processing components
        self.global_context = nn.Sequential(
            nn.Linear(hidden_size, hidden_size),
            nn.LayerNorm(hidden_size),
            nn.GELU()
        )
        
        self.spatial_attention = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 2),
            nn.GELU(),
            nn.Linear(hidden_size // 2, 1),
            nn.Sigmoid()
        )

        self.feature_refinement = nn.Sequential(
            nn.Linear(hidden_size, hidden_size),
            nn.LayerNorm(hidden_size),
            nn.Dropout(config.DROPOUT),
            nn.GELU()
        )

    def forward(self, pixel_values):
        # Forward pass with automatic device handling
        outputs = self.vit(pixel_values=pixel_values)
        vision_features = outputs.last_hidden_state
        
        # Feature enhancement
        global_repr = self.global_context(vision_features[:, 0:1, :])
        attn_weights = self.spatial_attention(vision_features)
        attn_weights = attn_weights / (attn_weights.sum(dim=1, keepdim=True) + 1e-6)
        
        attended_features = vision_features * attn_weights
        refined_features = self.feature_refinement(attended_features)
        enhanced_features = refined_features + global_repr
        
        return enhanced_features
```

```python




# Advanced Multimodal Fusion with Co-Attention and Transformer Integration
class AdvancedMultimodalFusion(nn.Module):
    def __init__(self, vision_dim, text_dim, fusion_dim, num_heads=8, num_layers=3, dropout=0.1):
        super(AdvancedMultimodalFusion, self).__init__()
        
        # Project features to common space
        self.vision_proj = nn.Linear(vision_dim, fusion_dim)
        self.text_proj = nn.Linear(text_dim, fusion_dim)
        
        # Co-Attention mechanisms
        self.co_attn_v2t = nn.MultiheadAttention(fusion_dim, num_heads, dropout=dropout)
        self.co_attn_t2v = nn.MultiheadAttention(fusion_dim, num_heads, dropout=dropout)
        
        # Transformer layers for deep fusion
        self.fusion_transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=fusion_dim,
                nhead=num_heads,
                dim_feedforward=fusion_dim * 4,
                dropout=dropout,
                activation='gelu',
                batch_first=True
            ),
            num_layers=num_layers
        )
        
        # Gated fusion with highway connections
        self.gate = nn.Sequential(
            nn.Linear(fusion_dim * 2, fusion_dim),
            nn.Sigmoid()
        )
        
        self.fusion_layer = nn.Linear(fusion_dim * 2, fusion_dim)
        self.norm1 = nn.LayerNorm(fusion_dim)
        
        # Enhanced FFN with Gated Linear Units
        self.ffn = nn.Sequential(
            nn.Linear(fusion_dim, fusion_dim * 4),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(fusion_dim * 4, fusion_dim),
            nn.Dropout(dropout)
        )
        
        self.norm2 = nn.LayerNorm(fusion_dim)
        
        # Additional MLP for feature enhancement
        self.feature_enhancement = nn.Sequential(
            nn.Linear(fusion_dim, fusion_dim),
            nn.LayerNorm(fusion_dim),
            nn.Dropout(dropout * 0.5),  # Lower dropout
            nn.GELU()
        )
    
    def forward(self, vision_features, text_features, text_attention_mask=None):
        batch_size = vision_features.size(0)
        
        # Project to common space
        vision_proj = self.vision_proj(vision_features)
        text_proj = self.text_proj(text_features)
        
        # Create attention mask for vision (all tokens attended to)
        vision_attention_mask = torch.ones(
            (batch_size, vision_features.size(1)),
            device=vision_features.device
        )
        
        # Co-attention: vision attending to text
        v2t_attn_mask = None
        if text_attention_mask is not None:
            v2t_attn_mask = ~text_attention_mask.bool()
        
        v2t_context, _ = self.co_attn_v2t(
            query=vision_proj.transpose(0, 1),
            key=text_proj.transpose(0, 1),
            value=text_proj.transpose(0, 1),
            key_padding_mask=v2t_attn_mask
        )
        v2t_context = v2t_context.transpose(0, 1)  # Batch first
        
        # Co-attention: text attending to vision
        t2v_context, _ = self.co_attn_t2v(
            query=text_proj.transpose(0, 1),
            key=vision_proj.transpose(0, 1),
            value=vision_proj.transpose(0, 1)
        )
        t2v_context = t2v_context.transpose(0, 1)  # Batch first
        
        # Concatenate original features with cross-attended features
        enhanced_vision = torch.cat([vision_proj, v2t_context], dim=-1)
        enhanced_text = torch.cat([text_proj, t2v_context], dim=-1)
        
        # Gated fusion for vision
        gate_v = self.gate(enhanced_vision)
        fused_v = self.fusion_layer(enhanced_vision)
        fused_v = gate_v * fused_v + (1 - gate_v) * vision_proj  # Highway connection
        
        # Gated fusion for text
        gate_t = self.gate(enhanced_text)
        fused_t = self.fusion_layer(enhanced_text)
        fused_t = gate_t * fused_t + (1 - gate_t) * text_proj  # Highway connection
        
        # Create joint representation
        # Use the [CLS] token from text and vision
        text_cls = fused_t[:, 0:1, :]  # (batch_size, 1, fusion_dim)
        vision_cls = fused_v[:, 0:1, :]  # (batch_size, 1, fusion_dim)
        
        # Combine them
        combined_cls = torch.cat([text_cls, vision_cls], dim=1)  # (batch_size, 2, fusion_dim)
        
        # Process through transformer for deep fusion
        # Create mask that allows attention only between tokens (no padding)
        cls_attention_mask = torch.ones((batch_size, 2), device=combined_cls.device).bool()
        
        # Apply transformer
        transformer_output = self.fusion_transformer(combined_cls)
        
        # Pool the outputs (use the first token as the representation)
        pooled_output = transformer_output[:, 0, :]
        
        # Final feature enhancement
        enhanced_output = self.feature_enhancement(pooled_output)
        
        return enhanced_output.unsqueeze(1)  # (batch_size, 1, fusion_dim)



```

```python
class AdvancedVQAModel(nn.Module):
    def __init__(self, num_answers, config):
        super().__init__()
        
        # Store config
        self.config = config
        
        # Visual encoder
        self.visual_encoder = AdvancedVisionFeatures(config)
        vision_dim = self.visual_encoder.vit.config.hidden_size
        
        # Text encoder
        self.text_encoder = BertModel.from_pretrained(config.BERT_MODEL)
        text_dim = self.text_encoder.config.hidden_size

        # Disable gradient checkpointing for TPU compatibility
        if config.USE_GRADIENT_CHECKPOINTING and not TPU_MODE:
            self.visual_encoder.vit.gradient_checkpointing_enable()
            self.text_encoder.gradient_checkpointing_enable()
        else:
            if hasattr(self.visual_encoder.vit, "gradient_checkpointing_disable"):
                self.visual_encoder.vit.gradient_checkpointing_disable()
            if hasattr(self.text_encoder, "gradient_checkpointing_disable"):
                self.text_encoder.gradient_checkpointing_disable()

        # Fusion module
        self.fusion = AdvancedMultimodalFusion(
            vision_dim=vision_dim,
            text_dim=text_dim,
            fusion_dim=config.FUSION_OUTPUT_SIZE,
            num_heads=16,
            num_layers=4,
            dropout=config.DROPOUT
        )

        # Classifier components
        self.classifier_heads = nn.ModuleList([
            nn.Sequential(
                nn.Linear(config.FUSION_OUTPUT_SIZE, config.FUSION_OUTPUT_SIZE),
                nn.LayerNorm(config.FUSION_OUTPUT_SIZE),
                nn.GELU(),
                nn.Dropout(config.DROPOUT),
                nn.Linear(config.FUSION_OUTPUT_SIZE, config.FUSION_OUTPUT_SIZE // 2),
                nn.GELU(),
                nn.Dropout(config.DROPOUT * 0.5),
                nn.Linear(config.FUSION_OUTPUT_SIZE // 2, num_answers)
            ) for _ in range(3)
        ])

        self.classifier_fusion = nn.Parameter(torch.ones(3) / 3)
        self.question_type_classifier = nn.Linear(config.FUSION_OUTPUT_SIZE, 7)
        self._init_weights()

    def _init_weights(self):
        for head in self.classifier_heads:
            for module in head:
                if isinstance(module, nn.Linear):
                    nn.init.xavier_uniform_(module.weight)
                    if module.bias is not None:
                        nn.init.constant_(module.bias, 0)

    def forward(self,image, pixel_values, input_ids, attention_mask):
    # Process in smaller chunks if on TPU
        if TPU_MODE:
            batch_size = pixel_values.size(0)
            chunk_size = 2  # Process 2 samples at a time
            outputs = []
            
            for i in range(0, batch_size, chunk_size):
                # Get chunk
                end_idx = min(i + chunk_size, batch_size)
                chunk_pixel_values = pixel_values[i:end_idx]
                chunk_input_ids = input_ids[i:end_idx]
                chunk_attention_mask = attention_mask[i:end_idx]
                
                # Process chunk
                with torch.no_grad():  # Reduce memory usage during forward pass
                    chunk_vision_features = self.visual_encoder(chunk_pixel_values)
                    chunk_text_outputs = self.text_encoder(
                        input_ids=chunk_input_ids,
                        attention_mask=chunk_attention_mask
                    )
                    chunk_text_features = chunk_text_outputs.last_hidden_state
                
                chunk_fused_features = self.fusion(
                    vision_features=chunk_vision_features,
                    text_features=chunk_text_features,
                    text_attention_mask=chunk_attention_mask
                )
                
                chunk_fused_cls = chunk_fused_features[:, 0, :]
                chunk_head_logits = [head(chunk_fused_cls) for head in self.classifier_heads]
                
                outputs.append((chunk_head_logits, chunk_fused_cls))
                
                # Force XLA execution
                if TPU_MODE:
                    xm.mark_step()
            
            # Combine results
            all_head_logits = [[torch.cat([out[0][i] for out in outputs], dim=0) for i in range(3)]]
            all_fused_cls = torch.cat([out[1] for out in outputs], dim=0)
            
        else:
            # Original forward pass for GPU/CPU
            vision_features = self.visual_encoder(pixel_values)
            text_outputs = self.text_encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
            text_features = text_outputs.last_hidden_state
            
            fused_features = self.fusion(
                vision_features=vision_features,
                text_features=text_features,
                text_attention_mask=attention_mask
            )
            
            fused_cls = fused_features[:, 0, :]
            all_head_logits = [head(fused_cls) for head in self.classifier_heads]
            all_fused_cls = fused_cls
        
        # Common processing for both paths
        fusion_weights = F.softmax(self.classifier_fusion, dim=0)
        logits = sum(w * logit for w, logit in zip(fusion_weights, all_head_logits[0]))
        
        return {
            'logits': logits,
            'head_logits': all_head_logits[0],
            'q_type_logits': self.question_type_classifier(all_fused_cls),
            'fusion_weights': fusion_weights
        }
            
```

```python
def mixup_data(x, y, alpha=0.2):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size(0)
    index = torch.randperm(batch_size).to(x.device)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

```

```python
def train_one_epoch(model, dataloader, optimizer, criterion, device, epoch, config, scaler=None):
    model.train()
    progress_bar = tqdm(dataloader, desc=f'Training Epoch {epoch+1}')
    
    total_loss = 0
    correct_predictions = 0
    total_predictions = 0
    
    # Default weights if not in config
    head_loss_weight = getattr(config, 'HEAD_LOSS_WEIGHT', 0.1)  # Default to 0.1 if not specified
    qtype_loss_weight = getattr(config, 'QTYPE_LOSS_WEIGHT', 0.1)  # Default to 0.1 if not specified
    
    for step, batch in enumerate(progress_bar):
        # Move batch to device and prepare inputs
        inputs = {
            'image': batch['image'].to(device),
            'pixel_values': batch['image'].to(device),
            'input_ids': batch['input_ids'].to(device),
            'attention_mask': batch['attention_mask'].to(device)
        }
        
        # Check for the correct label key
        if 'answer' in batch:
            labels = batch['answer'].to(device)
        elif 'answers' in batch:
            labels = batch['answers'].to(device)
        elif 'target' in batch:
            labels = batch['target'].to(device)
        elif 'targets' in batch:
            labels = batch['targets'].to(device)
        elif 'label' in batch:
            labels = batch['label'].to(device)
        else:
            raise KeyError(f"No label key found in batch. Available keys: {batch.keys()}")
        
        # Clear gradients
        optimizer.zero_grad()
        
        # Forward pass with unpacked dictionary
        outputs = model(**inputs)
        
        # Calculate loss
        logits = outputs['logits']
        loss = criterion(logits, labels)
        
        # Add auxiliary losses if available and enabled
        if 'head_logits' in outputs and head_loss_weight > 0:
            head_losses = [criterion(head_logit, labels) for head_logit in outputs['head_logits']]
            head_loss = sum(head_losses) / len(head_losses)
            loss = loss + head_loss_weight * head_loss
        
        if 'q_type_logits' in outputs and 'question_types' in batch and qtype_loss_weight > 0:
            q_type_loss = F.cross_entropy(outputs['q_type_logits'], batch['question_types'].to(device))
            loss = loss + qtype_loss_weight * q_type_loss
        
        # Backward pass
        loss.backward()
        
        # Gradient clipping if enabled
        if hasattr(config, 'GRADIENT_CLIPPING') and config.GRADIENT_CLIPPING > 0:
            torch.nn.utils.clip_grad_norm_(model.parameters(), config.GRADIENT_CLIPPING)
        
        # Optimizer step with TPU handling
        if TPU_MODE:
            xm.optimizer_step(optimizer)
            xm.mark_step()
        else:
            optimizer.step()
        
        # Calculate accuracy
        predictions = torch.argmax(logits, dim=-1)
        correct_predictions += (predictions == labels).sum().item()
        total_predictions += labels.size(0)
        
        # Update total loss
        total_loss += loss.item()
        
        # Update progress bar
        current_lr = optimizer.param_groups[0]['lr']
        progress_bar.set_postfix({
            'loss': f'{loss.item():.4f}',
            'acc': f'{100.0 * correct_predictions / total_predictions:.2f}%',
            'lr': f'{current_lr:.6f}'
        })
        
        # Memory management for TPU
        if TPU_MODE and step % 100 == 0:
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
    
    # Calculate epoch metrics
    avg_loss = total_loss / len(dataloader)
    accuracy = 100.0 * correct_predictions / total_predictions
    
    return {
        'train_loss': avg_loss,
        'train_accuracy': accuracy
    }
```

```python
def validate(model, dataloader, criterion, device, config):
    model.eval()
    progress_bar = tqdm(dataloader, desc='Validation')
    
    total_loss = 0.0
    correct_predictions = 0
    total_predictions = 0
    
    with torch.no_grad():
        for batch in progress_bar:
            inputs = {
                'image': batch['image'].to(device),
                'pixel_values': batch['image'].to(device),
                'input_ids': batch['input_ids'].to(device),
                'attention_mask': batch['attention_mask'].to(device)
            }
            # Determine labels key in batch
            if 'answer' in batch:
                labels = batch['answer'].to(device)
            elif 'answers' in batch:
                labels = batch['answers'].to(device)
            elif 'target' in batch:
                labels = batch['target'].to(device)
            elif 'targets' in batch:
                labels = batch['targets'].to(device)
            elif 'label' in batch:
                labels = batch['label'].to(device)
            else:
                raise KeyError(f"No label key found in batch. Available keys: {batch.keys()}")
            
            outputs = model(**inputs)
            logits = outputs['logits']
            loss = criterion(logits, labels)
            total_loss += loss.item()
            
            predictions = torch.argmax(logits, dim=-1)
            correct_predictions += (predictions == labels).sum().item()
            total_predictions += labels.size(0)
            
            progress_bar.set_postfix({
                'loss': f"{loss.item():.4f}",
                'acc': f"{100.0 * correct_predictions / total_predictions:.2f}%"
            })
            
            if TPU_MODE and total_predictions % 1000 == 0:
                gc.collect()
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                    
    avg_loss = total_loss / len(dataloader)
    accuracy = 100.0 * correct_predictions / total_predictions if total_predictions > 0 else 0
    
    return {
        'val_loss': avg_loss,
        'val_accuracy': accuracy
    }
```

```python
def process_batch_in_chunks(model, pixel_values, input_ids, attention_mask, chunk_size=2):
    """
    Process a batch in smaller chunks to avoid memory issues on TPU.
    Last updated: 2025-03-24 06:36:34 by DARKRAI1234
    """
    batch_size = pixel_values.size(0)
    outputs_list = []
    
    for i in range(0, batch_size, chunk_size):
        end_idx = min(i + chunk_size, batch_size)
        # For TPU, we need to access the underlying model
        if TPU_MODE:
            chunk_outputs = model._model(
                pixel_values=pixel_values[i:end_idx],
                input_ids=input_ids[i:end_idx],
                attention_mask=attention_mask[i:end_idx]
            )
        else:
            chunk_outputs = model(
                pixel_values=pixel_values[i:end_idx],
                input_ids=input_ids[i:end_idx],
                attention_mask=attention_mask[i:end_idx]
            )
        outputs_list.append(chunk_outputs)
        
        # Force XLA compilation
        if TPU_MODE:
            xm.mark_step()
    
    # Combine outputs
    combined_outputs = {
        'logits': torch.cat([o['logits'] for o in outputs_list], dim=0),
        'head_logits': [torch.cat([o['head_logits'][i] for o in outputs_list], dim=0) for i in range(len(outputs_list[0]['head_logits']))],
        'q_type_logits': torch.cat([o['q_type_logits'] for o in outputs_list], dim=0),
        'fusion_weights': outputs_list[0]['fusion_weights']
    }
    
    return combined_outputs
```

```python
def get_scheduler(optimizer, config, num_training_steps):
    """Create scheduler based on config"""
    warmup_steps = int(num_training_steps * config.WARMUP_RATIO)
    
    if config.SCHEDULER == "linear":
        return get_cosine_schedule_with_warmup(
            optimizer,
            num_warmup_steps=warmup_steps,
            num_training_steps=num_training_steps
        )
    elif config.SCHEDULER == "cosine_with_restarts":
        # Custom cosine with restarts
        class CosineAnnealingWarmupRestarts(optim.lr_scheduler._LRScheduler):
            def __init__(self, optimizer, T_0, T_mult=1, eta_max=0.1, T_up=0, gamma=1., last_epoch=-1):
                self.T_0 = T_0
                self.T_mult = T_mult
                self.eta_max = eta_max
                self.T_up = T_up
                self.T_i = T_0
                self.gamma = gamma
                self.cycle = 0
                self.T_cur = last_epoch
                super(CosineAnnealingWarmupRestarts, self).__init__(optimizer, last_epoch)

            def get_lr(self):
                if self.T_cur == -1:
                    return [base_lr for base_lr in self.base_lrs]
                elif self.T_cur < self.T_up:
                    return [base_lr + (self.eta_max - base_lr) * self.T_cur / self.T_up for base_lr in self.base_lrs]
                else:
                    return [base_lr + (self.eta_max - base_lr) * 
                            (1 + math.cos(math.pi * (self.T_cur - self.T_up) / (self.T_i - self.T_up))) / 2
                            for base_lr in self.base_lrs]

            def step(self, epoch=None):
                if epoch is None:
                    epoch = self.last_epoch + 1
                    self.T_cur = self.T_cur + 1
                    if self.T_cur >= self.T_i:
                        self.cycle += 1
                        self.T_cur = self.T_cur - self.T_i
                        self.T_i = (self.T_i - self.T_up) * self.T_mult + self.T_up
                else:
                    self.T_cur = epoch
                self.last_epoch = math.floor(epoch)
                self.eta_max = self.eta_max * (self.gamma ** self.cycle)
                self._last_lr = [group['lr'] for group in self.optimizer.param_groups]
                for param_group, lr in zip(self.optimizer.param_groups, self.get_lr()):
                    param_group['lr'] = lr
        
        # Calculate cycle length
        cycle_length = num_training_steps // config.NUM_RESTARTS
        
        return CosineAnnealingWarmupRestarts(
            optimizer,
            T_0=cycle_length,
            T_mult=1,
            eta_max=config.LEARNING_RATE,
            T_up=warmup_steps,
            gamma=0.9  # Reduce max LR by 10% each cycle
        )
    else:
        # Default to linear warmup then constant
        return get_cosine_schedule_with_warmup(
            optimizer,
            num_warmup_steps=warmup_steps,
            num_training_steps=num_training_steps
        )

# Function to log metrics to wandb
def log_metrics(metrics, epoch, phase="train"):
    """Log metrics to wandb and console"""
    # Print to console
    metric_str = f"Epoch {epoch+1} {phase} metrics: "
    for k, v in metrics.items():
        if isinstance(v, dict):
            continue  # Skip nested dictionaries for console output
        if isinstance(v, list):
            continue  # Skip lists for console output
        metric_str += f"{k}: {v:.4f}, "
    
    logger.info(metric_str)
    
    # Log to wandb if enabled
    if wandb.run is not None:
        wandb_metrics = {}
        for k, v in metrics.items():
            if isinstance(v, dict):
                # Handle nested dictionaries
                for sub_k, sub_v in v.items():
                    wandb_metrics[f"{phase}_{k}_{sub_k}"] = sub_v
            elif isinstance(v, list):
                # Handle lists
                for i, item in enumerate(v):
                    wandb_metrics[f"{phase}_{k}_{i}"] = item
            else:
                wandb_metrics[f"{phase}_{k}"] = v
        
        wandb_metrics["epoch"] = epoch
        wandb.log(wandb_metrics)

```

```python
def train_model(model, train_dataloader, val_dataloader, config):
    logger.info("Starting model training...")
    
    # Optimizer setup
    no_decay = ['bias', 'LayerNorm.weight']
    optimizer_grouped_parameters = [
        {
            'params': [p for n, p in model.named_parameters() 
                      if not any(nd in n for nd in no_decay)],
            'weight_decay': config.WEIGHT_DECAY,
        },
        {
            'params': [p for n, p in model.named_parameters() 
                      if any(nd in n for nd in no_decay)],
            'weight_decay': 0.0,
        },
    ]
    
    optimizer = torch.optim.AdamW(optimizer_grouped_parameters, 
                                lr=config.LEARNING_RATE)
    
    # Loss function
    criterion = torch.nn.CrossEntropyLoss(
        label_smoothing=config.LABEL_SMOOTHING, 
        ignore_index=-1
    )
    criterion = CustomCrossEntropyLoss()
    # Scheduler
    total_steps = len(train_dataloader) * config.NUM_EPOCHS // config.GRADIENT_ACCUMULATION_STEPS
    scheduler = get_scheduler(optimizer, config, total_steps)
    
    # Checkpoint handling
    start_epoch = 0
    best_val_accuracy = 0
    if config.RESUME_FROM_CHECKPOINT:
        checkpoint = torch.load(config.RESUME_FROM_CHECKPOINT)
        model.load_state_dict(checkpoint['model_state_dict'])
        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        if scheduler and checkpoint['scheduler_state_dict']:
            scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
        start_epoch = checkpoint['epoch']
        best_val_accuracy = checkpoint['best_val_accuracy']
        logger.info(f"Resuming from checkpoint: Epoch {start_epoch}, "
                  f"Best Val Acc: {best_val_accuracy:.4f}")
    
    # Training variables
    patience_counter = 0
    os.makedirs(config.CHECKPOINT_DIR, exist_ok=True)
    
    # Training loop
    for epoch in range(start_epoch, config.NUM_EPOCHS):
        logger.info(f"Epoch {epoch+1}/{config.NUM_EPOCHS}")
        
        # Train phase
        train_metrics = train_one_epoch(
            model=model,
            dataloader=train_dataloader,
            optimizer=optimizer,
            criterion=criterion,
            scaler=GradScaler(enabled=config.USE_AMP),
            device=config.DEVICE,
            epoch=epoch,
            config=config
        )
        
        # Validation phase
        val_metrics, _, _ = validate(model, val_dataloader, criterion, config.DEVICE)
        
        # Update learning rate
        scheduler.step()
        
        # Checkpoint saving
        current_val_accuracy = val_metrics['val_accuracy']
        is_best = current_val_accuracy > best_val_accuracy
        best_val_accuracy = max(best_val_accuracy, current_val_accuracy)
        
        # Save checkpoint every epoch
        checkpoint = {
            'epoch': epoch + 1,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'scheduler_state_dict': scheduler.state_dict(),
            'best_val_accuracy': best_val_accuracy,
            'config': config.__dict__,
            'val_metrics': val_metrics
        }
        
        # Regular checkpoint
        checkpoint_path = os.path.join(
            config.CHECKPOINT_DIR, 
            f'checkpoint_epoch_{epoch+1}.pth'
        )
        torch.save(checkpoint, checkpoint_path)
        logger.info(f"Saved checkpoint to {checkpoint_path}")
        
        # Best checkpoint
        if is_best:
            best_path = os.path.join(config.CHECKPOINT_DIR, 'best_model.pth')
            torch.save(checkpoint, best_path)
            logger.info(f"New best model saved with accuracy {best_val_accuracy:.4f}")
            patience_counter = 0
        else:
            patience_counter += 1
            logger.info(f"Val accuracy didn't improve. Patience {patience_counter}/{config.PATIENCE}")
            
            if patience_counter >= config.PATIENCE:
                logger.info("Early stopping triggered!")
                break
        
        # Memory cleanup
        if not TPU_MODE and torch.cuda.is_available():
            torch.cuda.empty_cache()
    
    logger.info(f"Training complete. Best validation accuracy: {best_val_accuracy:.4f}")
    return model, best_val_accuracy
```

```python
def evaluate_model(model, val_dataloader, answer_space, idx_to_answer, config, num_examples=10):
    model.eval()
    
    all_predictions = []
    all_ground_truth = []
    example_results = []
    
    with torch.no_grad():
        for i, batch in enumerate(tqdm(val_dataloader, desc="Evaluating")):
            # Move batch to device
            pixel_values = batch['pixel_values'].to(config.DEVICE)
            input_ids = batch['input_ids'].to(config.DEVICE)
            attention_mask = batch['attention_mask'].to(config.DEVICE)
            labels = batch['label'].to(config.DEVICE)
            
            # Forward pass
            outputs = model(pixel_values, input_ids, attention_mask)
            logits = outputs['logits']
            
            # Get predictions
            preds = torch.argmax(logits, dim=1).cpu().numpy()
            
            # Get confidence scores (softmax)
            confidence = F.softmax(logits, dim=1).max(dim=1)[0].cpu().numpy()
            
            # Convert to answer text
            pred_answers = [idx_to_answer[pred] for pred in preds]
            true_answers = [idx_to_answer[label.item()] for label in labels.cpu()]
            
            all_predictions.extend(pred_answers)
            all_ground_truth.extend(true_answers)
            
            # Store some examples for visualization
            if len(example_results) < num_examples:
                for j in range(min(len(batch['image_id']), num_examples - len(example_results))):
                    example_results.append({
                        'image_id': batch['image_id'][j],
                        'question': batch['question'][j],
                        'true_answer': true_answers[j],
                        'pred_answer': pred_answers[j],
                        'confidence': confidence[j],
                        'correct': pred_answers[j] == true_answers[j]
                    })
    
    # Calculate overall accuracy
    accuracy = sum(p == t for p, t in zip(all_predictions, all_ground_truth)) / len(all_predictions)
    logger.info(f"Overall accuracy: {accuracy:.4f}")
    
    # Calculate accuracy by answer length
    length_groups = {
        'short': {'correct': 0, 'total': 0},
        'medium': {'correct': 0, 'total': 0},
        'long': {'correct': 0, 'total': 0}
    }
    
    for pred, true in zip(all_predictions, all_ground_truth):
        length = len(true.split())
        if length <= 1:
            group = 'short'
        elif length <= 3:
            group = 'medium'
        else:
            group = 'long'
        
        length_groups[group]['total'] += 1
        if pred == true:
            length_groups[group]['correct'] += 1
    
    logger.info("Accuracy by answer length:")
    for group, stats in length_groups.items():
        if stats['total'] > 0:
            group_acc = stats['correct'] / stats['total']
            logger.info(f"  {group}: {group_acc:.4f} ({stats['correct']}/{stats['total']})")
    
    # Print example results
    logger.info("\nExample predictions:")
    for ex in example_results:
        status = "✓" if ex['correct'] else "✗"
        logger.info(f"Q: {ex['question']}")
        logger.info(f"True: {ex['true_answer']} | Pred: {ex['pred_answer']} ({ex['confidence']:.4f}) {status}")
        logger.info("-" * 40)
    
    return accuracy, example_results

# Error analysis for incorrect predictions
def analyze_errors(model, val_dataloader, idx_to_answer, config):
    model.eval()
    
    # Categorize errors
    error_categories = {
        'synonyms': [],  # Similar but not exact match
        'antonyms': [],  # Opposite meaning
        'related': [],   # Related but incorrect
        'unrelated': []  # Completely unrelated
    }
    
    # Common antonym pairs for error analysis
    antonym_pairs = [
        ('left', 'right'), ('up', 'down'), ('top', 'bottom'),
        ('front', 'back'), ('white', 'black'), ('yes', 'no'),
        ('small', 'large'), ('light', 'dark'), ('open', 'closed')
    ]
    
    with torch.no_grad():
        for batch in tqdm(val_dataloader, desc="Analyzing errors"):
            # Move batch to device
            pixel_values = batch['pixel_values'].to(config.DEVICE)
            input_ids = batch['input_ids'].to(config.DEVICE)
            attention_mask = batch['attention_mask'].to(config.DEVICE)
            labels = batch['label'].to(config.DEVICE)
            
            # Forward pass
            outputs = model(pixel_values, input_ids, attention_mask)
            logits = outputs['logits']
            
            # Get top 5 predictions
            top_k_values, top_k_indices = torch.topk(logits, k=5, dim=1)
            top_k_probs = F.softmax(top_k_values, dim=1).cpu().numpy()
            top_k_preds = top_k_indices.cpu().numpy()
            
            # Convert to answer text
            top_k_answers = [
                [idx_to_answer[pred] for pred in preds]
                for preds in top_k_preds
            ]
            
            # Get true answers
            true_labels = labels.cpu().numpy()
            true_answers = [idx_to_answer[label] for label in true_labels]
            
            # Check for errors
            for i, (true_answer, pred_answers, probs) in enumerate(zip(true_answers, top_k_answers, top_k_probs)):
                if pred_answers[0] != true_answer:  # Top prediction is wrong
                    # Get the question for context
                    question = batch['question'][i]
                    image_id = batch['image_id'][i]
                    
                    # Check error category
                    error_info = {
                        'question': question,
                        'image_id': image_id,
                        'true_answer': true_answer,
                        'pred_answer': pred_answers[0],
                        'pred_prob': probs[0],
                        'top5_answers': pred_answers,
                        'top5_probs': probs.tolist()
                    }
                    
                    # Check for antonyms
                    is_antonym = False
                    for a, b in antonym_pairs:
                        if (true_answer.lower() == a and pred_answers[0].lower() == b) or \
                           (true_answer.lower() == b and pred_answers[0].lower() == a):
                            error_categories['antonyms'].append(error_info)
                            is_antonym = True
                            break
                    
                    if is_antonym:
                        continue
                    
                    # Check for synonyms or related terms
                    # This is a simplified check - in real applications, use a proper NLP approach
                    if true_answer.lower() in pred_answers[0].lower() or pred_answers[0].lower() in true_answer.lower():
                        error_categories['synonyms'].append(error_info)
                    elif true_answer in pred_answers[1:] or any(true_answer.lower() in pa.lower() for pa in pred_answers[1:]):
                        error_categories['related'].append(error_info)
                    else:
                        error_categories['unrelated'].append(error_info)
    
    # Print error analysis summary
    logger.info("\nError Analysis:")
    for category, errors in error_categories.items():
        logger.info(f"{category}: {len(errors)} errors")
        
        # Print some examples
        if errors:
            logger.info("Examples:")
            for i, error in enumerate(errors[:3]):  # Show first 3 examples
                logger.info(f"Q: {error['question']}")
                logger.info(f"True: {error['true_answer']} | Pred: {error['pred_answer']} ({error['pred_prob']:.4f})")
                logger.info(f"Top 5: {', '.join([f'{a} ({p:.2f})' for a, p in zip(error['top5_answers'], error['top5_probs'])])}")
                logger.info("-" * 40)
    
    return error_categories
```

```python
def ensemble_predictions(models, dataloader, idx_to_answer, config):
    """Combine predictions from multiple models"""
    for model in models:
        model.eval()
    
    all_predictions = []
    all_ground_truth = []
    
    with torch.no_grad():
        for batch in tqdm(dataloader, desc="Ensemble prediction"):
            # Move batch to device
            pixel_values = batch['pixel_values'].to(config.DEVICE)
            input_ids = batch['input_ids'].to(config.DEVICE)
            attention_mask = batch['attention_mask'].to(config.DEVICE)
            labels = batch['label'].cpu().numpy()
            
            # Get predictions from all models
            ensemble_logits = None
            for model in models:
                outputs = model(pixel_values, input_ids, attention_mask)
                logits = outputs['logits']
                
                if ensemble_logits is None:
                    ensemble_logits = logits.cpu().numpy()
                else:
                    ensemble_logits += logits.cpu().numpy()
            
            # Average the logits
            ensemble_logits /= len(models)
            
            # Get predictions
            preds = np.argmax(ensemble_logits, axis=1)
            
            # Convert to answer text
            pred_answers = [idx_to_answer[pred] for pred in preds]
            true_answers = [idx_to_answer[label] for label in labels]
            
            all_predictions.extend(pred_answers)
            all_ground_truth.extend(true_answers)
    
    # Calculate accuracy
    accuracy = sum(p == t for p, t in zip(all_predictions, all_ground_truth)) / len(all_predictions)
    logger.info(f"Ensemble model accuracy: {accuracy:.4f}")
    
    return accuracy, all_predictions, all_ground_truth
```

```python
def manage_checkpoints(new_checkpoint_path):
    """
    Creates a checkpoint directory in /kaggle/working/my_checkpoints if it doesn't exist, 
    then stores the new .pth file there. If there are more than 2 .pth files 
    in that directory, the oldest file is deleted.
    
    Args:
         new_checkpoint_path (str): the file path to the new checkpoint file.
    """
    checkpoint_dir = "/kaggle/working/my_checkpoints"
    
    if not os.path.exists(checkpoint_dir):
        os.makedirs(checkpoint_dir)
        logger.info(f"Created directory {checkpoint_dir}")
    
    # List all .pth files in the directory
    pth_files = [os.path.join(checkpoint_dir, f) for f in os.listdir(checkpoint_dir) if f.endswith(".pth")]
    
    # If there are more than or equal to 2 checkpoint files, remove the oldest one
    if len(pth_files) >= 2:
        pth_files.sort(key=lambda x: os.path.getmtime(x))
        oldest_file = pth_files[0]
        os.remove(oldest_file)
        logger.info(f"Deleted oldest checkpoint: {oldest_file}")
    
    # Copy the new checkpoint file into the checkpoint directory
    basename = os.path.basename(new_checkpoint_path)
    destination = os.path.join(checkpoint_dir, basename)
    shutil.copy(new_checkpoint_path, destination)
    logger.info(f"Stored new checkpoint in working directory: {destination}")
```

```python
def load_checkpoint(checkpoint_path, model, optimizer, scheduler, device):
    """
    Load checkpoint with proper PyTorch 2.6+ compatibility
    """
    try:
        logger.info(f"Loading checkpoint from: {checkpoint_path}")
        
        if not os.path.exists(checkpoint_path):
            raise FileNotFoundError(f"Checkpoint file not found at: {checkpoint_path}")
        
        # Allow numpy scalar in safe globals
        from torch.serialization import add_safe_globals
        import numpy as np
        add_safe_globals([np.ndarray, np._core.multiarray.scalar])
        
        # Try loading with weights_only=False since we trust our own checkpoint
        checkpoint = torch.load(checkpoint_path, map_location='cpu', weights_only=False)
        logger.info("Successfully loaded checkpoint")
        
        # Load model state with error handling
        try:
            model_state = checkpoint['model_state_dict']
            if TPU_MODE:
                if not any(k.startswith('module.') for k in model_state.keys()) and \
                   any(k.startswith('module.') for k in model.state_dict().keys()):
                    model_state = {'module.' + k: v for k, v in model_state.items()}
                elif any(k.startswith('module.') for k in model_state.keys()) and \
                     not any(k.startswith('module.') for k in model.state_dict().keys()):
                    model_state = {k.replace('module.', ''): v for k, v in model_state.items()}
            
            model.load_state_dict(model_state)
            logger.info("Model state loaded successfully")
        except Exception as e:
            logger.error(f"Error loading model state: {str(e)}")
            raise
        
        # Load optimizer state
        try:
            optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
            logger.info("Optimizer state loaded successfully")
        except Exception as e:
            logger.error(f"Error loading optimizer state: {str(e)}")
            raise
        
        # Load scheduler state if it exists
        if scheduler and checkpoint.get('scheduler_state_dict'):
            try:
                scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
                logger.info("Scheduler state loaded successfully")
            except Exception as e:
                logger.warning(f"Error loading scheduler state: {str(e)}")
        
        start_epoch = checkpoint['epoch']
        metrics = checkpoint['metrics']
        
        logger.info(f"Resumed from epoch {start_epoch} with metrics:")
        logger.info(f"Training Loss: {metrics['train_loss']:.4f}")
        logger.info(f"Validation Loss: {metrics['val_loss']:.4f}")
        logger.info(f"Validation Accuracy: {metrics['val_accuracy']:.4f}")
        
        return start_epoch, metrics
        
    except Exception as e:
        logger.error("=" * 80)
        logger.error("Failed to load checkpoint:")
        logger.error(str(e))
        logger.error(traceback.format_exc())
        logger.error("=" * 80)
        raise
```

```python
def show_predictions(model, dataloader, device, config, idx_to_answer, num_examples=5):
    model.eval()
    examples_shown = 0
    plt.figure(figsize=(15, num_examples * 3))
    
    with torch.no_grad():
        for batch in dataloader:
            inputs = {
                'image': batch['image'].to(device),
                'pixel_values': batch['image'].to(device),
                'input_ids': batch['input_ids'].to(device),
                'attention_mask': batch['attention_mask'].to(device)
            }
            outputs = model(**inputs)
            logits = outputs['logits']
            predictions = torch.argmax(logits, dim=-1)
            
            # Optional question text from the batch; defaulting to "N/A"
            questions = batch.get('question_text', ["N/A"] * len(predictions))
            # Ground truth answers; assumes ground truth indices are stored under the "answer" key.
            actual_answers = batch.get('answer', None)
            images = batch['image'].cpu()  # assuming images are in CHW format
            
            for i in range(len(predictions)):
                if examples_shown >= num_examples:
                    break
                
                # Get predicted label text using idx_to_answer mapping.
                predicted_idx = predictions[i].item()
                predicted_label = idx_to_answer.get(predicted_idx, str(predicted_idx))
                
                # Get actual answer label text if available.
                if actual_answers is not None:
                    actual_idx = actual_answers[i].item() if torch.is_tensor(actual_answers[i]) else actual_answers[i]
                    actual_label = idx_to_answer.get(actual_idx, str(actual_idx))
                else:
                    actual_label = "N/A"
                    
                plt.subplot(num_examples, 1, examples_shown + 1)
                image = images[i].permute(1, 2, 0).numpy()  # Convert from CHW to HWC
                plt.imshow(image)
                plt.axis('off')
                plt.title(f"Question: {questions[i]} | Actual: {actual_label} | Predicted: {predicted_label}")
                examples_shown += 1
            if examples_shown >= num_examples:
                break

    plt.tight_layout()
    plt.show()
```

```python
def save_checkpoint(epoch, model, optimizer, scheduler, metrics, config, is_best, run_dir, keep_last_n=2):
    """
    Save a checkpoint of the training state, keeping only the last N checkpoints and the best model
    Last updated: 2025-03-24 07:09:49 by DARKRAI1234
    """
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model._model.state_dict() if TPU_MODE else model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict() if scheduler else None,
        'metrics': metrics,
        'config': config.__dict__
    }
    
    # Save the regular checkpoint
    checkpoint_path = os.path.join(run_dir, f'checkpoint_epoch_{epoch:03d}.pth')
    
    if TPU_MODE:
        xm.save(checkpoint, checkpoint_path)
    else:
        torch.save(checkpoint, checkpoint_path)
    logger.info(f"Saved checkpoint to {checkpoint_path}")
    
    # Save best model if this is the best so far
    if is_best:
        best_path = os.path.join(run_dir, 'best_model.pth')
        if TPU_MODE:
            xm.save(checkpoint, best_path)
        else:
            torch.save(checkpoint, best_path)
        logger.info(f"Saved best model to {best_path}")
    
    # Remove old checkpoints, keeping only the last N and the best model
    try:
        checkpoints = [f for f in os.listdir(run_dir) if f.startswith('checkpoint_epoch_') and f.endswith('.pth')]
        checkpoints.sort(key=lambda x: int(x.split('_')[-1].split('.')[0]))  # Sort by epoch number
        
        if len(checkpoints) > keep_last_n:
            checkpoints_to_delete = checkpoints[:-keep_last_n]  # Keep the last N checkpoints
            for old_checkpoint in checkpoints_to_delete:
                old_checkpoint_path = os.path.join(run_dir, old_checkpoint)
                try:
                    os.remove(old_checkpoint_path)
                    logger.info(f"Removed old checkpoint: {old_checkpoint}")
                except Exception as e:
                    logger.warning(f"Failed to remove old checkpoint {old_checkpoint}: {str(e)}")
    except Exception as e:
        logger.warning(f"Error while cleaning up old checkpoints: {str(e)}")
```
